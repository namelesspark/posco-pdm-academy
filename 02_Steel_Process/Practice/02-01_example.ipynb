```python
# STEP 3
# 측정 샘플 불러오기 (timestamp가 들어있는 파일)
SAMPLE_PATH = "..\\Data\\02-01_측정의_3요소_측정샘플.csv"
df_ts = pd.read_csv(SAMPLE_PATH, encoding="utf-8")

print(df_ts.info())

# 문자열을 시간 타입으로 바꿔야 뺄셈이 된다
df_ts["timestamp"] = pd.to_datetime(df_ts["timestamp"])

# 연속 타임스탬프의 차이로 실제 저장 간격 확인
간격 = df_ts["timestamp"].diff()
print(간격.dt.total_seconds().value_counts(dropna=False))
```

```python
# STEP 3 - 반복값과 최소 변화 폭
rows = []
for col in df_ts.columns[1:]:
    rows.append(
        {
            "태그명": col,
            "최솟값": df_ts[col].min(),
            "최댓값": df_ts[col].max(),
            "평균": round(df_ts[col].mean(), 2),
            "값이 변한 최소 폭": round(df_ts[col].diff().abs().replace(0, pd.NA).min(), 4),
        }
    )

step3 = pd.DataFrame(rows)
print(step3.to_markdown(index=False))
```

```python
# 행간 데이터 변화 중 최소 폭은 메서드로 알 수 없습니다

# 최소폭을 구하는 로직을 만들어서 제출해주세요

# 파일로 제출하지 말고 텍스트 복사/붙여넣기로 진행해주세요.
for col in df_ts.columns[1:]:
    block = (df_ts[col] != df_ts[col].shift()).cumsum()  # 값이 바뀔 때마다 번호 증가
    max_repeat = df_ts.groupby(block)[col].size().max()
    min_step = df_ts[col].diff().abs().replace(0, pd.NA).min()  # 0(변화없음)은 제외
    res = df_hyd_fur.set_index("tag")["resolution"].get(col)
    print(
        f"{col:16} 최대 연속반복 {max_repeat:2}회 | 최소 변화폭 {min_step} | 분해능 {res}"
    )
```
